```python
import pandas as pd
import numpy as np

df=pd.read_excel(r"C:\Users\HP\Desktop\marrketing.xlsx")

# 1. الأداء حسب المنصة والهدف
print("1. Performance Metrics by Platform and Objective")
platform_objective_pivot = pd.pivot_table(
    df,
    values=['Clicks', 'Conversions', 'Budget (USD)'],
    index='Platform',
    columns='Objective',
    aggfunc={'Clicks': 'sum', 'Conversions': 'sum', 'Budget (USD)': 'sum'},
    fill_value=0,
    margins=True
)
print(platform_objective_pivot)

# 2. كفاءة التحويل حسب نوع الحملة
print("\n2. Conversion Efficiency by Campaign Type")
conversion_efficiency_pivot = pd.pivot_table(
    df,
    values=['Conversions', 'Budget (USD)'],
    index='Campaign Type',
    aggfunc={'Conversions': 'sum', 'Budget (USD)': 'sum'},
    fill_value=0
)

# معالجة القسمة على الصفر
conversion_efficiency_pivot['Cost per Conversion'] = np.where(
    conversion_efficiency_pivot['Conversions'] > 0,
    conversion_efficiency_pivot['Budget (USD)'] / conversion_efficiency_pivot['Conversions'],
    0  # أو np.nan إذا كنت تفضل
)
print(conversion_efficiency_pivot)

# 3. تحليل الأداء الإقليمي
print("\n3. Regional Performance Analysis")
regional_pivot = pd.pivot_table(
    df,
    values=['Clicks', 'Conversions', 'Budget (USD)', 'Impressions'],
    index='Target Region',
    aggfunc={'Clicks': 'sum', 'Conversions': 'sum', 'Budget (USD)': 'sum', 'Impressions': 'sum'},
    fill_value=0
)

# حساب CTR ومعدل التحويل مع معالجة الأخطاء
regional_pivot['CTR (%)'] = np.where(
    regional_pivot['Impressions'] > 0,
    (regional_pivot['Clicks'] / regional_pivot['Impressions']) * 100,
    0
)

regional_pivot['Conversion Rate (%)'] = np.where(
    regional_pivot['Clicks'] > 0,
    (regional_pivot['Conversions'] / regional_pivot['Clicks']) * 100,
    0
)
print(regional_pivot)

# 4. أداء المنتج حسب المنصة
print("\n4. Product Performance by Platform")
product_platform_pivot = pd.pivot_table(
    df,
    values=['Conversions', 'Budget (USD)'],
    index='Product Promoted',
    columns='Platform',
    aggfunc={'Conversions': 'sum', 'Budget (USD)': 'sum'},
    fill_value=0,
    margins=True
)
print(product_platform_pivot)

# 5. أداء الحملة الشهرية
print("\n5. Monthly Campaign Performance")
# تحويل التاريخ بشكل صحيح
df['Start Date'] = pd.to_datetime(df['Start Date'])
df['Start Month'] = df['Start Date'].dt.to_period('M')

monthly_pivot = pd.pivot_table(
    df,
    values=['Clicks', 'Conversions', 'Budget (USD)'],
    index='Start Month',
    columns='Campaign Type',
    aggfunc={'Clicks': 'sum', 'Conversions': 'sum', 'Budget (USD)': 'sum'},
    fill_value=0
)
print(monthly_pivot)

# 6. تحليل عائد الاستثمار حسب نوع الحملة والمنتج (بديل أبسط)
print("\n6. ROI Analysis by Campaign Type and Product")

# استخدام groupby بدلاً من pivot_table لتجنب مشاكل MultiIndex
roi_grouped = df.groupby(['Campaign Type', 'Product Promoted']).agg({
    'Conversions': 'sum',
    'Budget (USD)': 'sum'
}).reset_index()

# حساب عائد الاستثمار
conversion_value = 50  # قيمة كل تحويل
roi_grouped['ROI (%)'] = np.where(
    roi_grouped['Budget (USD)'] > 0,
    (roi_grouped['Conversions'] * conversion_value - roi_grouped['Budget (USD)']) / roi_grouped['Budget (USD)'] * 100,
    0
)

print(roi_grouped[['Campaign Type', 'Product Promoted', 'ROI (%)']])
```

```python
import pandas as pd
import numpy as np
from dash import Dash, dcc, html, Input, Output
import plotly.express as px
import plotly.graph_objects as go

# ==============================
# تحميل البيانات ومعالجة القيم
# ==============================
df=pd.read_excel(r"C:\Users\HP\Desktop\marrketing.xlsx")
df['Start Month'] = pd.to_datetime(df['Start Date']).dt.to_period('M').astype(str)

# ==============================
# حساب القيم الأساسية (KPIs)
# ==============================
def calc_kpis(filtered_df):
    اجمالي_كل_الحملات = len(filtered_df)
    اجمالي_ميزانية_الحملات = filtered_df['Budget (USD)'].sum()
    اجمالي_مرات_الظهور = filtered_df['Impressions'].sum()
    اجمالي_النقرات = filtered_df['Clicks'].sum()
    اجمالي_التحويلات = filtered_df['Conversions'].sum()
    معدل_النقر_الى_الظهور = (اجمالي_النقرات / اجمالي_مرات_الظهور) * 100 if اجمالي_مرات_الظهور else 0
    معدل_التحويل = (اجمالي_التحويلات / اجمالي_النقرات) * 100 if اجمالي_النقرات else 0
    متوسط_تكلفة_النقرة = اجمالي_ميزانية_الحملات / اجمالي_النقرات if اجمالي_النقرات else 0
    متوسط_تكلفة_التحويل = اجمالي_ميزانية_الحملات / اجمالي_التحويلات if اجمالي_التحويلات else 0

    return {
        "total_campaigns": اجمالي_كل_الحملات,
        "total_budget": اجمالي_ميزانية_الحملات,
        "total_impressions": اجمالي_مرات_الظهور,
        "total_clicks": اجمالي_النقرات,
        "total_conversions": اجمالي_التحويلات,
        "ctr": معدل_النقر_الى_الظهور,
        "conversion_rate": معدل_التحويل,
        "cpc": متوسط_تكلفة_النقرة,
        "cpa": متوسط_تكلفة_التحويل
    }

# ==============================
# إنشاء التطبيق
# ==============================
app = Dash(__name__)

app.layout = html.Div([
    html.H1("📊 لوحة تحكم الحملات الإعلانية", style={"textAlign": "center"}),

    # ======== Slicers ========
    html.Div([
        html.Div([
            html.Label("اختر المنصة:"),
            dcc.Dropdown(
                options=[{"label": i, "value": i} for i in df["Platform"].unique()],
                multi=True,
                id="platform-filter"
            )
        ]),
        html.Div([
            html.Label("اختر نوع الحملة:"),
            dcc.Dropdown(
                options=[{"label": i, "value": i} for i in df["Campaign Type"].unique()],
                multi=True,
                id="campaign-type-filter"
            )
        ]),
        html.Div([
            html.Label("اختر الشهر:"),
            dcc.Dropdown(
                options=[{"label": i, "value": i} for i in df["Start Month"].unique()],
                multi=True,
                id="month-filter"
            )
        ]),
    ], style={"display": "grid", "gridTemplateColumns": "repeat(3, 1fr)", "gap": "15px"}),

    # ======== KPI Cards ========
    html.Div(id="kpi-cards", style={
        "display": "grid",
        "gridTemplateColumns": "repeat(4, 1fr)",
        "gap": "15px",
        "marginTop": "20px"
    }),

    # ======== الرسوم البيانية ========
    html.Div([
        dcc.Graph(id="platform-objective-chart"),
        dcc.Graph(id="monthly-performance-chart"),
    ], style={"display": "grid", "gridTemplateColumns": "1fr 1fr", "gap": "20px"}),

    html.Div([
        dcc.Graph(id="regional-performance-chart"),
        dcc.Graph(id="budget-pie-chart"),
    ], style={"display": "grid", "gridTemplateColumns": "1fr 1fr", "gap": "20px"}),

    html.Div([
        dcc.Graph(id="ctr-gauge"),
        dcc.Graph(id="conversion-gauge"),
    ], style={"display": "grid", "gridTemplateColumns": "1fr 1fr", "gap": "20px"})
])

# ==============================
# Callback للتحديث التفاعلي
# ==============================
@app.callback(
    [Output("kpi-cards", "children"),
     Output("platform-objective-chart", "figure"),
     Output("monthly-performance-chart", "figure"),
     Output("regional-performance-chart", "figure"),
     Output("budget-pie-chart", "figure"),
     Output("ctr-gauge", "figure"),
     Output("conversion-gauge", "figure")],
    [Input("platform-filter", "value"),
     Input("campaign-type-filter", "value"),
     Input("month-filter", "value")]
)
def update_dashboard(selected_platforms, selected_campaign_types, selected_months):
    filtered_df = df.copy()
    if selected_platforms:
        filtered_df = filtered_df[filtered_df["Platform"].isin(selected_platforms)]
    if selected_campaign_types:
        filtered_df = filtered_df[filtered_df["Campaign Type"].isin(selected_campaign_types)]
    if selected_months:
        filtered_df = filtered_df[filtered_df["Start Month"].isin(selected_months)]

    kpis = calc_kpis(filtered_df)

    # ===== Cards =====
    def card(title, value, color="#1f77b4"):
        return html.Div([
            html.H4(title, style={"textAlign": "center", "color": color}),
            html.H3(value, style={"textAlign": "center", "fontWeight": "bold"})
        ], style={"background": "#f9f9f9", "padding": "10px", "borderRadius": "10px",
                  "boxShadow": "0 2px 5px rgba(0,0,0,0.1)"})

    kpi_cards = [
        card("📊 إجمالي الحملات", kpis['total_campaigns']),
        card("💰 إجمالي الميزانية", f"${kpis['total_budget']:,.2f}", "#2ca02c"),
        card("👁️ مرات الظهور", f"{kpis['total_impressions']:,}", "#9467bd"),
        card("🖱️ النقرات", f"{kpis['total_clicks']:,}", "#d62728"),
        card("🎯 التحويلات", f"{kpis['total_conversions']:,}", "#ff7f0e"),
        card("📈 CTR", f"{kpis['ctr']:.2f}%", "#17becf"),
        card("🔁 معدل التحويل", f"{kpis['conversion_rate']:.2f}%", "#bcbd22"),
        card("💵 CPC", f"${kpis['cpc']:.4f}", "#8c564b"),
        card("🏷️ CPA", f"${kpis['cpa']:.4f}", "#e377c2"),
    ]

    # ===== رسم 1: أداء المنصة والهدف =====
    fig1 = px.bar(filtered_df, x="Platform", y="Clicks", color="Objective",
                  title="📊 الأداء حسب المنصة والهدف", barmode="group")

    # ===== رسم 2: الأداء الشهري (Line Chart بدلاً من أعمدة) =====
    monthly_pivot = filtered_df.groupby("Start Month")[["Clicks", "Conversions"]].sum().reset_index()
    fig2 = go.Figure()
    fig2.add_trace(go.Scatter(x=monthly_pivot["Start Month"], y=monthly_pivot["Clicks"], name="Clicks", mode="lines+markers"))
    fig2.add_trace(go.Scatter(x=monthly_pivot["Start Month"], y=monthly_pivot["Conversions"], name="Conversions", mode="lines+markers"))
    fig2.update_layout(title="📅 الأداء الشهري", xaxis_title="الشهر", yaxis_title="القيمة")

    # ===== رسم 3: الأداء الإقليمي =====
    regional_pivot = filtered_df.groupby("Target Region")[["Clicks", "Conversions"]].sum().reset_index()
    fig3 = px.bar(regional_pivot, x="Target Region", y=["Clicks", "Conversions"],
                  title="🌍 الأداء الإقليمي", barmode="group")

    # ===== رسم 4: توزيع الميزانية (Pie Chart) =====
    budget_pivot = filtered_df.groupby("Campaign Type")["Budget (USD)"].sum().reset_index()
    fig4 = px.pie(budget_pivot, names="Campaign Type", values="Budget (USD)", title="🧾 توزيع الميزانية حسب نوع الحملة")

    # ===== Gauge Charts =====
    fig_ctr = go.Figure(go.Indicator(
        mode="gauge+number", value=kpis['ctr'], title={'text': "CTR (%)"},
        gauge={'axis': {'range': [0, max(10, kpis['ctr']*1.5)]}, 'bar': {'color': "blue"}}
    ))

    fig_conv = go.Figure(go.Indicator(
        mode="gauge+number", value=kpis['conversion_rate'], title={'text': "Conversion Rate (%)"},
        gauge={'axis': {'range': [0, max(10, kpis['conversion_rate']*1.5)]}, 'bar': {'color': "green"}}
    ))

    return kpi_cards, fig1, fig2, fig3, fig4, fig_ctr, fig_conv


if __name__ == "__main__":
    app.run(debug=True)
```
